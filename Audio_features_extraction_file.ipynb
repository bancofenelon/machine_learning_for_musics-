```python
import os
import warnings
import librosa
import numpy as np
import pandas as pd
from  features_extraction import extraire_features
warnings.filterwarnings('ignore')
```

```python
# --- CONFIGURATION ---
# Chemin vers le dossier racine contenant les 10 sous-dossiers de genres
#dossier_musique = r"C:\Users\Habila\Desktop\Fenelon\Data\genres_original" 
dossier_musique = r"C:\Users\ADMIN\gtzanProject\audio_data"
fichier_csv_sortie = "my_gtzan_features.csv"
limite_fichiers = 1000

```

```python


# BOUCLE PRINCIPALE
print("--- DÉMARRAGE DE L'EXTRACTION DES FEATURES ---")
donnees_musiques = []
fichiers_traites = 0

for racine, _, fichiers in os.walk(dossier_musique):
    for nom_fichier in fichiers:
        if fichiers_traites >= limite_fichiers: break
        if not nom_fichier.lower().endswith(('.wav', '.mp3', '.flac', '.m4a')):
            continue
            
        print(f"[{fichiers_traites + 1}/{limite_fichiers}] Extraction depuis : {nom_fichier}...")
        try:
            extraction_features = extraire_features(os.path.join(racine, nom_fichier))
            if extraction_features is not None:
                extraction_features.update({'nom_fichier': nom_fichier, 'genre': os.path.basename(racine)})
                donnees_musiques.append(extraction_features)
        except Exception as e:
            print(f"  -> Erreur sur {nom_fichier}: {str(e)}")
        except:
            print("Erreur")

        fichiers_traites += 1    
    if fichiers_traites >= limite_fichiers:
        break
```

```python

# --- SAUVEGARDE ET EXPORT ---
print("\n--- EXTRACTION TERMINÉE ---")
print("Conversion en DataFrame Pandas et sauvegarde...")
# Création du DataFrame final
df = pd.DataFrame(donnees_musiques)
# Sauvegarde en fichier CSV sans l'index numérique de Pandas
df.to_csv(fichier_csv_sortie, index=False, encoding='utf-8')
print(f"Base de données sauvegardée sous : {fichier_csv_sortie}")
print(f"Dimensions du dataset : {df.shape[0]} lignes, {df.shape[1]} colonnes.")
print("\nAperçu des 5 premières lignes :")
print(df.shape)
df.head(2)

```
